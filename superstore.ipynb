```python
if date_cols:
    df['year'] = df[date_cols[0]].dt.year
    df['month'] = df[date_cols[0]].dt.month
    df['month_name'] = df[date_cols[0]].dt.month_name()

df.head()
```

```python
sales_cols = [c for c in df.columns if 'sal' in c.lower() or 'amount' in c.lower() or 'revenue' in c.lower()]
profit_cols = [c for c in df.columns if 'profit' in c.lower()]

print("Sales candidates:", sales_cols)
print("Profit candidates:", profit_cols)
```

```python
if sales_cols and profit_cols:
    sales_col = sales_cols[0]
    profit_col = profit_cols[0]

    total_sales = df[sales_col].sum()
    total_profit = df[profit_col].sum()

    print("Total Sales:", total_sales)
    print("Total Profit:", total_profit)

    print("Profit Margin %:", (total_profit / total_sales) * 100)
else:
    print("Could not find proper sales/profit columns")
```

```python
region_cols = [c for c in df.columns if 'region' in c.lower()]

if region_cols and sales_cols:
    df.groupby(region_cols[0])[sales_cols[0]].sum().plot(kind='bar')
else:
    print("Region column not available")
```

```python
category_cols = [c for c in df.columns if 'category' in c.lower()]

if category_cols and sales_cols:
    df.groupby(category_cols[0])[sales_cols[0]].sum().sort_values().plot(kind='bar')
    import matplotlib.pyplot as plt
    plt.title("Sales by Category")
    plt.show()
else:
    print("Category column not found")
```

```python
subcat_cols = [c for c in df.columns if 'sub' in c.lower()]

if subcat_cols and sales_cols:
    df.groupby(subcat_cols[0])[sales_cols[0]].sum().sort_values().plot(kind='barh')
    import matplotlib.pyplot as plt
    plt.title("Sales by Sub-Category")
    plt.show()
else:
    print("Sub-category column not found")
```

```python
product_cols = [c for c in df.columns if 'product' in c.lower() or 'name' in c.lower()]

if product_cols and sales_cols:
    df.groupby(product_cols[0])[sales_cols[0]].sum().sort_values(ascending=False).head(10).plot(kind='barh')
    import matplotlib.pyplot as plt
    plt.title("Top 10 Products by Sales")
    plt.show()
else:
    print("Product column not found")
```

```python
if product_cols and profit_cols:
    df.groupby(product_cols[0])[profit_cols[0]].sum().sort_values().head(10).plot(kind='barh', color='red')
    import matplotlib.pyplot as plt
    plt.title("Top Loss Making Products")
    plt.show()
else:
    print("Profit/Product columns not found")
```

```python
if 'month_name' in df.columns and sales_cols:
    df.groupby('month_name')[sales_cols[0]].sum().plot(kind='line', marker='o')
    import matplotlib.pyplot as plt
    plt.title("Monthly Sales Trend")
    plt.show()
else:
    print("Monthly data not available")
```
